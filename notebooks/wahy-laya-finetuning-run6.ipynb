# Laya run 6: Envelope v3 — The Clean, Pruned Architecture (Kaggle, GPU T4 ×2)

Sixth run, testing **Envelope v3 (Pruned v2 + Golden v1 Layout)**.

**Why this run exists:**
* **Run 1 (`soft_asis` baseline):** Outstanding verdict quality (**0.944 AUC**, **91.4% accuracy**, **6.4% false rejections**),
  but occasionally dropped verses in the middle of a continuous recitation chain (e.g. الواقعة 89).
* **Run 3 & 4 (Envelope v2):** Merging consecutive verses into passages completely eliminated dropped chain verses (0 missed),
  **BUT** the v2 envelope introduced too many noisy features at once:
  1. Shuffled `canonical_corpus_text` to the very bottom behind hundreds of transcript tokens.
  2. Injected noisy `neighbor_findings` raw JSON blobs.
  This diluted cross-attention and caused verdict AUC to collapse from **0.944 down to 0.836**!
* **Run 5:** Restoring corpus text to the top helped structure, but training *exclusively* on merged passages degraded
  single-verse sensitivity (false rejections jumped to **14.0%**, short-quote accuracy fell to **58.8%**).

### The "Envelope v3" Architecture:
1. **Corpus text at TOP (v1 proven order):** `source_type` -> `candidate_reference` -> `canonical_corpus_text`.
2. **Noise completely eliminated:** Drop `neighbor_findings` raw JSON blobs.
3. **Keep lexical boundary indicators:** `immediate_transcript_before`, `unmatched_corpus_prefix_words`, `prefix_words_heard`,
   `immediate_transcript_after`, `unmatched_corpus_suffix_words`, `suffix_words_heard` (helps cut-off recognition without noise).
4. **Deduplication:** Ayah text is only in `canonical_corpus_text` (no duplicate in context).
5. **Max length:** 2048 tokens.

| run | data | envelope | questions | what it answers |
|---|---|---|---|---|
| `v3_single` | Single verses | **Envelope v3** (clean cues, corpus at top) | verdict only | Does the clean v3 envelope preserve/beat ~0.94 AUC on single verses? |
| `v3_mixed` | Single verses + passages | **Envelope v3** (clean cues, corpus at top) | verdict only | Does mixing single verses with passages give passage continuity WITHOUT degrading single-verse accuracy? |

**Benchmarks to beat:**
* `before: soft_asis` (production model): 91.4% accuracy, 0.944 AUC, 6.4% real rejected.
* Run 4 `v1_verdict`: 91.4% accuracy, 0.933 AUC, 4.1% real rejected.

## 3. Dataset, scripts, base checkpoint, Hub login

```python
import glob, json, sys
from pathlib import Path
from huggingface_hub import snapshot_download, login
from laya.agent import _fix_tokenizer_config

DATA = next(Path(p).parent for p in sorted(glob.glob("/kaggle/input/**/manifest.json", recursive=True))
            if json.loads(Path(p).read_text()).get("name") == "laya_v2")
manifest = json.loads((DATA / "manifest.json").read_text())
sys.path.insert(0, str(DATA))
import eval_laya

print("dataset:", DATA, "| created", manifest.get("created"), "| base model", manifest.get("base_model"))

WORK = Path("/kaggle/working")
for d in ("results", "logs", "data"):
    (WORK / d).mkdir(parents=True, exist_ok=True)

# Builder for Envelope v3 (Corpus at TOP, clean boundary indicators, NO neighbor_findings)
def to_v3_state(st):
    return {
        "source_type": st.get("source_type"),
        "candidate_reference": st.get("candidate_reference"),
        "canonical_corpus_text": st.get("canonical_corpus_text"),  # Golden placement: AT TOP
        "has_citation_cue": bool(st.get("has_citation_cue", False)),
        "context_before": st.get("context_before"),
        "matched_span_in_lecture": st.get("matched_span_in_lecture"),
        "context_after": st.get("context_after"),
        "immediate_transcript_before": st.get("immediate_transcript_before", ""),
        "unmatched_corpus_prefix_words": st.get("unmatched_corpus_prefix_words", ""),
        "prefix_words_heard": st.get("prefix_words_heard", ""),
        "immediate_transcript_after": st.get("immediate_transcript_after", ""),
        "unmatched_corpus_suffix_words": st.get("unmatched_corpus_suffix_words", ""),
        "suffix_words_heard": st.get("suffix_words_heard", ""),
        "quran_neighboring_context": st.get("quran_neighboring_context")
    }

v1_instructions = (
    "In this Arabic Islamic lecture transcript, determine whether `matched_span_in_lecture` "
    "(surrounded by `context_before` and `context_after`) is an actual quotation/citation of "
    "`canonical_corpus_text` (`source_type`), or if it is merely ordinary speaker prose/explanation "
    "that happens to share words with the corpus text."
)
v1_criteria = {
    "CORRECT": "The speaker is genuinely quoting, reciting, or invoking this Quranic Ayah, Prophetic Hadith, or Dhikr (verbatim or as a direct religious citation in context).",
    "FALSE": "False positive: the words are ordinary conversational speech, fiqh explanation, or narrator/person names that coincidentally overlap with corpus vocabulary without quoting the text.",
    "NOT_SURE": "Ambiguous or partial paraphrase/meaning-only allusion where it is borderline whether the speaker intended a direct quotation."
}

def convert_file(in_path, out_path):
    out_path.parent.mkdir(parents=True, exist_ok=True)
    with open(in_path, "r", encoding="utf-8") as fin, open(out_path, "w", encoding="utf-8") as fout:
        for line in fin:
            r = json.loads(line)
            row = {
                "id": r["id"],
                "state": to_v3_state(r["state"]),
                "questions": {"verdict": {"type": "choice", "instructions": v1_instructions, "criteria": v1_criteria}},
                "gold": {"verdict": r["gold"]["verdict"]},
                "meta": r["meta"]
            }
            fout.write(json.dumps(row, ensure_ascii=False) + "\n")

# 1. Build v3_single from DATA/base (Single verses in Envelope v3)
V3_SINGLE_DIR = WORK / "data" / "v3_single"
if not (V3_SINGLE_DIR / "train.jsonl").exists():
    print("Building v3_single dataset (single verses, clean boundary cues, corpus at top)...")
    for split in ("train", "val", "test"):
        out_f = V3_SINGLE_DIR / f"{split}.jsonl"
        convert_file(DATA / "base" / f"{split}.jsonl", out_f)
        print(f"  v3_single {split}: {out_f.stat().st_size / 1e3:.0f} KB")

# 2. Build v3_mixed (Single verses + multi-verse passages combined)
V3_MIXED_DIR = WORK / "data" / "v3_mixed"
if not (V3_MIXED_DIR / "train.jsonl").exists():
    print("Building v3_mixed dataset (single verses + merged passages combined)...")
    for split in ("train", "val", "test"):
        V3_MIXED_DIR.mkdir(parents=True, exist_ok=True)
        single_rows = [json.loads(l) for l in open(V3_SINGLE_DIR / f"{split}.jsonl", encoding="utf-8")]
        seen_ids = {r["id"] for r in single_rows}
        multi_passages = []
        for l in open(DATA / "passages" / f"{split}.jsonl", encoding="utf-8"):
            r = json.loads(l)
            if r["meta"].get("verses", 1) > 1 and r["id"] not in seen_ids:
                row = {
                    "id": r["id"],
                    "state": to_v3_state(r["state"]),
                    "questions": {"verdict": {"type": "choice", "instructions": v1_instructions, "criteria": v1_criteria}},
                    "gold": {"verdict": r["gold"]["verdict"]},
                    "meta": r["meta"]
                }
                multi_passages.append(row)
        combined = single_rows + multi_passages
        with open(V3_MIXED_DIR / f"{split}.jsonl", "w", encoding="utf-8") as fout:
            for r in combined:
                fout.write(json.dumps(r, ensure_ascii=False) + "\n")
        print(f"  v3_mixed {split}: {len(combined)} rows ({len(single_rows)} single + {len(multi_passages)} multi-verse passages)")

MODEL_ID = manifest["base_model"]
model_dir = snapshot_download(MODEL_ID)
_fix_tokenizer_config(model_dir)
print("base checkpoint:", model_dir)

try:
    from kaggle_secrets import UserSecretsClient
    HF_TOKEN = UserSecretsClient().get_secret("HF_TOKEN")
    login(token=HF_TOKEN)
    print("Hub: logged in")
except Exception as e:
    HF_TOKEN = None
    print("Hub: no HF_TOKEN secret:", e)
```

## 4. Settings
Both models train on **Verdict Only** with 2 epochs, effective batch 64 (4 × 8 accumulation), max_len 2048, and RLCD loss.

```python
RUNS = {
    "v3_single": {
        "data_dir": V3_SINGLE_DIR,
        "questions": ["verdict"],
        "cfg_extra": {"wahy_envelope": 3, "wahy_envelope_options": {"passages": False}, "wahy_questions": ["verdict"]},
        "about": "Envelope v3: corpus at top, clean boundary indicators, single verses only"
    },
    "v3_mixed": {
        "data_dir": V3_MIXED_DIR,
        "questions": ["verdict"],
        "cfg_extra": {"wahy_envelope": 3, "wahy_envelope_options": {"passages": True}, "wahy_questions": ["verdict"]},
        "about": "Envelope v3: corpus at top, clean boundary indicators, single verses + passages combined"
    }
}

TRAIN_ARGS = (f"--epochs 2 --micro-batch 4 --grad-accum 8 --lr-encoder 2.5e-5 --lr-head 1e-4 --loss rlcd "
              f"--max-len 2048 --head-max-len 256")
APP_FALSE_SHARE = 0.10
BEFORE_MODEL = "ZenoAbderrahim/laya-wahy-soft_asis"
PUSH_TO_HUB = True
HUB_PREFIX = "ZenoAbderrahim/laya-wahy-r6"

def run(cmd, log):
    print("$", cmd, flush=True)
    with open(log, "w") as f:
        p = subprocess.Popen(cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
        for line in p.stdout:
            f.write(line)
            if "Warning" not in line and "warn(" not in line:
                print(line, end="", flush=True)
        return p.wait()

import pandas as pd
rows = []
for name, cfg in RUNS.items():
    tr_lines = [json.loads(l) for l in open(cfg["data_dir"] / "train.jsonl")]
    n_rows = len(tr_lines)
    n_false = sum(r["meta"]["verdict"] == "FALSE" for r in tr_lines)
    rows.append({
        "run": name,
        "questions": ", ".join(cfg["questions"]),
        "about": cfg["about"],
        "train rows": n_rows,
        "not a quote (share)": round(n_false / n_rows, 3)
    })
pd.DataFrame(rows).set_index("run")
```

## 5. The "before" line: current production model

```python
flags = f"--verse-flags {DATA}/test_verse_flags.json"
if HF_TOKEN:
    rc = run(f"python {DATA}/eval_laya.py predict --model {BEFORE_MODEL} --name 'before: soft_asis (app)' "
             f"--test {DATA}/v1_test.jsonl --out {WORK}/results/before_soft_asis.json --app-false-share {APP_FALSE_SHARE} "
             f"--batch-size 32 --baselines {DATA}/v1_baselines.json {flags}",
             WORK / "logs" / "eval_before.log")
    print("before:", "ok" if rc == 0 else f"failed (exit {rc}), see logs/eval_before.log")
```

## 6. Train + evaluate

```python
import time
status = {}

for name, cfg in RUNS.items():
    data_dir = cfg["data_dir"]
    qs = cfg["questions"]
    out = WORK / "models" / name
    t0 = time.time()
    try:
        cfg_extra_clean = json.dumps(cfg["cfg_extra"]).replace('"', '\\"')
        qarg = f"--questions {','.join(qs)}" if qs else ""
        rc = run(f"torchrun --standalone --nproc_per_node=2 {DATA}/train_ddp.py {model_dir} {out} "
                 f"--train {data_dir}/train.jsonl --val {data_dir}/val.jsonl --name laya-wahy-r5-{name} "
                 f"--cfg-extra \"{cfg_extra_clean}\" {qarg} {TRAIN_ARGS}", WORK / "logs" / f"train_{name}.log")
        if rc != 0 or not (out / "model.safetensors").exists():
            status[name] = f"training failed (exit {rc}), see logs/train_{name}.log"
            continue
        
        tr_lines = [json.loads(l) for l in open(data_dir / "train.jsonl")]
        tr_false_share = round(sum(r["meta"]["verdict"] == "FALSE" for r in tr_lines) / len(tr_lines), 4)
        
        rc = run(f"python {DATA}/eval_laya.py predict --model {out} --name {name} --test {data_dir}/test.jsonl "
                 f"--out {WORK}/results/{name}.json --train-false-share {tr_false_share} "
                 f"--app-false-share {APP_FALSE_SHARE} --batch-size 16 {flags}", WORK / "logs" / f"eval_{name}.log")
        status[name] = f"ok in {(time.time() - t0) / 60:.0f} min" if rc == 0 else f"evaluation failed (exit {rc})"
        train_log = out / "train_log.json"
        if train_log.exists():
            (WORK / "results" / f"train_log_{name}.json").write_text(train_log.read_text())
    except Exception as e:
        status[name] = f"error: {e}"
    print(f"\n##### {name}: {status[name]}\n", flush=True)

json.dump(status, open(WORK / "results" / "status.json", "w"), indent=1)
status
```

## 9. Decision rule & comparison against benchmarks

```python
print("=== BENCHMARK COMPARISON (Cards: you) ===")
print("Run 1 (soft_asis production):  AUC 0.944 | Acc 91.4% | Real rejected 6.4%")
print("Run 2 (curve_100):             AUC 0.945 | Acc 89.4% | Real rejected 9.9%")
print("Run 4 (v1_verdict):            AUC 0.933 | Acc 91.4% | Real rejected 4.1%")
print("Run 5 (v1_passages):           AUC 0.924 | Acc 84.9% | Real rejected 14.0%")
print("========================================\n")

target_model = "v3_mixed" if "v3_mixed" in res else list(res.keys())[0]
if target_model in res:
    r_you = res[target_model]["groups"]["you"]["verdict"]
    auc_val = r_you["AUC P(not a quote)"]
    acc_val = 100.0 * r_you["accuracy"][0]
    rej_val = 100.0 * r_you["real quotes wrongly rejected"][0]
    print(f"{target_model} result on 'you': AUC {auc_val:.3f} | Acc {acc_val:.1f}% | Real rejected {rej_val:.1f}%")
    if auc_val >= 0.94 and acc_val >= 90.0 and rej_val <= 7.0:
        print(f"\nSUCCESS: {target_model} beat/matched benchmarks with low false rejection!")
        print("-> Ready to deploy in production!")
    else:
        print(f"\nNotice: AUC is {auc_val:.3f}, Acc is {acc_val:.1f}%, Real rejected {rej_val:.1f}%.")
```
